```python
import pandas as pd
import plotly.express as px
from pathlib import Path

data_dir = Path("data") / "processed"
nav_file_path = data_dir / "cleaned_02_nav_history.csv"

df = pd.read_csv(nav_file_path)

df['date'] = pd.to_datetime(df['date'])
df = df.groupby('date').mean()
df = df.asfreq('B').ffill()

print("Data successfully loaded and cleaned!")
```

```python
fig = px.line(df, y='nav', title='Daily NAV Trend (2022-2026)')

fig.add_vrect(x0="2023-01-01", x1="2023-12-31", fillcolor="green", opacity=0.1, annotation_text="2023 Bull Run")
fig.add_vrect(x0="2024-01-01", x1="2024-12-31", fillcolor="red", opacity=0.1, annotation_text="2024 Correction")

fig.show()
```

```python
from pathlib import Path

output_path = Path("reports/nav_trend_analysis.html")
output_path.parent.mkdir(exist_ok=True)
fig.write_html(output_path)

print(f"Report successfully saved at: {output_path}")
```

```python
import pandas as pd
import plotly.express as px
from pathlib import Path

category_file = Path("data/processed/cleaned_05_category_inflows.csv")
df_cat = pd.read_csv(category_file)

df_pivot = df_cat.pivot(index='category', columns='month', values='net_inflow_crore')

fig_heat = px.imshow(
    df_pivot, 
    labels=dict(x="Month", y="Category", color="Inflow (Cr)"),
    title="Mutual Fund Inflow Heatmap by Category",
    color_continuous_scale='Viridis'
)

output_path = Path("reports/category_inflow_heatmap.html")
output_path.parent.mkdir(exist_ok=True)
fig_heat.write_html(output_path)

fig_heat.show()
print(f"Heatmap saved at: {output_path}")
```

```python
import pandas as pd
import plotly.express as px
from pathlib import Path

investor_file = Path("data/processed/cleaned_08_investor_transactions.csv")
df_inv = pd.read_csv(investor_file)

age_data = df_inv['age_group'].value_counts().reset_index()
age_data.columns = ['age_group', 'count']

fig_age = px.pie(
    age_data, 
    values='count', 
    names='age_group', 
    title='Investor Distribution by Age Group'
)

output_path = Path("reports/age_group_distribution.html")
output_path.parent.mkdir(exist_ok=True)
fig_age.write_html(output_path)
fig_age.show()
print(f"Report saved at: {output_path}")
```

```python
gender_data = df_inv['gender'].value_counts().reset_index()
gender_data.columns = ['gender', 'count']

fig_gender = px.pie(
    gender_data, 
    values='count', 
    names='gender', 
    title='Investor Distribution by Gender',
    hole=0.4
)

output_path = Path("reports/gender_split.html")
fig_gender.write_html(output_path)
fig_gender.show()
print(f"Report saved at: {output_path}")
```

```python
import pandas as pd
import plotly.express as px
from pathlib import Path

holdings_file = Path("data/processed/cleaned_09_portfolio_holdings.csv")
df_holdings = pd.read_csv(holdings_file)

sector_data = df_holdings.groupby('sector')['weight_pct'].sum().reset_index()

fig_donut = px.pie(
    sector_data, 
    values='weight_pct', 
    names='sector', 
    title='Sector Allocation Weightage',
    hole=0.4
)

output_path = Path("reports/sector_allocation.html")
output_path.parent.mkdir(exist_ok=True)
fig_donut.write_html(output_path)

fig_donut.show()
print(f"Report saved at: {output_path}")
```

```python
import pandas as pd
import plotly.express as px
from pathlib import Path

nav_file = Path("data/processed/cleaned_02_nav_history.csv")
df_nav = pd.read_csv(nav_file)

df_pivot = df_nav.pivot(index='date', columns='amfi_code', values='nav')

corr_matrix = df_pivot.corr()

fig_corr = px.imshow(
    corr_matrix,
    text_auto=True,
    title='NAV Correlation Matrix',
    labels=dict(x="AMFI Code", y="AMFI Code", color="Correlation")
)

output_path = Path("reports/nav_correlation_matrix.html")
output_path.parent.mkdir(exist_ok=True)
fig_corr.write_html(output_path)

fig_corr.show()
print(f"Report saved at: {output_path}")
```
